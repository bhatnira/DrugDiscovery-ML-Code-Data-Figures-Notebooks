```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define marker shapes for each species
marker_shapes = {
    'Human Test': 'o',
    'Human Independent': 's',
    'Eel': 'X',
    'Mouse': '^',
    'Cow': 'D',
    'Ray': 'P',
    'Mosquito': 'v'
}

# Define colors for vertical lines
line_colors = {
    'Human Test': 'blue',
    'Human Independent': 'red'
}

# Set style for publication quality with an emphasis on colors
sns.set(style='whitegrid', font_scale=1.0)

# Set font to Helvetica, and fallback to DejaVu Sans if Helvetica is not found
font = 'Arial' if 'Arial' in plt.matplotlib.font_manager.findSystemFonts() else 'DejaVu Sans'

# Set font and make all fonts bold
plt.rcParams['font.family'] = font  # Set the font to Helvetica or DejaVu Sans
plt.rcParams['font.weight'] = 'bold'  # Make all fonts bold
plt.rcParams['axes.labelweight'] = 'bold'  # Make axis labels bold
plt.rcParams['axes.titlesize'] = 12  # Adjust title size
plt.rcParams['axes.labelsize'] = 12  # Adjust label size

# Initialize plot counter to handle the third plot differently
plot_counter = 0

# Iterate over each variant and create a combined plot for AUC ROC
for variant in data['Variant'].unique():
    plot_counter += 1  # Increment plot counter for each variant

    variant_data = data[data['Variant'] == variant]

    # Define figure size in inches (standard size for all plots)
    figsize = (5, 3)  # Default size for plots

    # For the third plot, make it slightly bigger
    if plot_counter == 3:
        figsize = (8, 6)  # Make the third plot slightly bigger

    # Create the plot
    fig, axes = plt.subplots(1, 1, figsize=figsize)

    # Create a smaller violin plot for AUC ROC with white filling
    sns.violinplot(ax=axes, data=variant_data, x='AUC ROC', y='Model Type', inner=None, linewidth=1.2, edgecolor='black', color='white', scale='count', width=0.9)
    axes.set_xlabel('AUC ROC', fontsize=12, fontweight='bold')  # Adjust font size for x-label
    axes.set_ylabel(f'{variant}', fontsize=12, fontweight='bold')  # Adjust font size for y-label

    handles = []  # to store legend handles
    labels = []   # to store legend labels

    for species, marker in marker_shapes.items():
        species_data = variant_data[variant_data['Species'] == species]
        if not species_data.empty:
            # Add stripplot to the violin plot for AUC ROC
            sns.stripplot(ax=axes, data=species_data, x='AUC ROC', y='Model Type', marker=marker, color='black', dodge=True, size=5, alpha=0.8)

            # Add species to legend
            if species not in labels:
                handles.append(plt.Line2D([0], [0], marker=marker, color='w', markerfacecolor='black', markersize=8))
                labels.append(species)

    # Calculate and add lines for top values of Human Test and Human Independent
    if 'Human Test' in variant_data['Species'].values:
        top_human_test_auc = variant_data[variant_data['Species'] == 'Human Test']['AUC ROC'].max()
        axes.axvline(x=top_human_test_auc, color=line_colors['Human Test'], linestyle='--', linewidth=1)

    if 'Human Independent' in variant_data['Species'].values:
        top_human_independent_auc = variant_data[variant_data['Species'] == 'Human Independent']['AUC ROC'].max()
        axes.axvline(x=top_human_independent_auc, color=line_colors['Human Independent'], linestyle='--', linewidth=1)

    # Add legends and place it very close to the plot (just below it)
    fig.legend(handles, labels, title='Species', loc='upper center', ncol=4, bbox_to_anchor=(0.5, -0.01), frameon=False, fontsize='small', markerscale=1.2)

    # Remove spines for a cleaner look
    sns.despine(ax=axes, top=True, right=True, bottom=True, left=True)

    # Adjust layout to bring the legend even closer
    plt.subplots_adjust(bottom=0.5)  # Reduce the space between the plot and the legend
    plt.tight_layout(pad=0.0, rect=[0, 0, 1, 0.95])  # Further reduce the bottom padding to ensure the legend fits closely

    # Save with high resolution in TIFF format
    plt.savefig(f'violin_plot_{variant}.tiff', format='tiff', dpi=300, bbox_inches='tight')
    plt.show()  # Close the figure to free up memory
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Assuming df is your DataFrame with columns: 'Feature', 'Test Accuracy', 'ROC-AUC', 'Type'

# Separate bar diagrams for Test Accuracy and ROC-AUC
features = classicalVsEnsembleVsAutoML['Feature'].unique()
types = classicalVsEnsembleVsAutoML['Type'].unique()

bar_width = 0.35  # Adjusted width for better separation
bar_positions = np.arange(len(features)) * (len(types) + 1) * bar_width  # Adjusted position for space between features

# Save each plot separately
for metric in ['Accuracy', 'ROC-AUC Score']:
    plt.figure(figsize=(10, 6))  # Adjust the size as needed

    for i, model_type in enumerate(types):
        subset_df = classicalVsEnsembleVsAutoML[classicalVsEnsembleVsAutoML['Type'] == model_type]
        plt.bar(bar_positions + i * bar_width, subset_df[metric], width=bar_width, label=f'{model_type} - {metric}')

    plt.xlabel('Descriptor/Fingerprint')
    plt.ylabel(metric)
    plt.title(f'{metric} for best models in different Descriptor/Fingerprint')
    plt.xticks(bar_positions + bar_width * (len(types) - 1) / 2, features)  # Adjusted position for better alignment

    # Remove box border around the plot
    plt.box(False)

    # Place the legend within the plot area using bbox_to_anchor
    plt.legend(loc='upper left', bbox_to_anchor=(1, 1), frameon=False)

    # Adjust layout for better spacing
    plt.tight_layout()

    # Save the individual figure as a high-resolution PNG file
    plt.savefig(f'{metric}_performance_plot.png', dpi=300, bbox_inches='tight')
    plt.show()
```

```python
from scipy import stats

model_groups = [group['ROC-AUC'].values for name, group in data.groupby('Model Type')]
f_stat, p_value = stats.f_oneway(*model_groups)

print(f"ANOVA F-statistic: {f_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("Significant difference between model types (p < 0.05)")
else:
    print("No significant difference between model types (p >= 0.05)")
```

```python
from PIL import Image

# Image paths for each row (three images per row)
image_paths_list = [
    [
        "....",
        "....",
        "...."
    ],
    [
        "....",
        "....",
        "...."
    ],
    [
        "....",
        "....",
        "...."
    ]
]

Image.MAX_IMAGE_PIXELS = None  # Prevent DecompressionBombError

dpi = 1200

# Smaller image size
target_width_inches = 2.5  # Previously 6
target_width_pixels = int(target_width_inches * dpi)

# Adjust height based on a fixed ratio (e.g., 2:1 width to height)
target_height_pixels = int(target_width_pixels * 0.5)

# Process each row
row_images = []
for image_paths in image_paths_list:
    imgs = [Image.open(p).convert("RGB").resize((target_width_pixels, target_height_pixels)) for p in image_paths]

    row_width = len(imgs) * target_width_pixels
    row_image = Image.new('RGB', (row_width, target_height_pixels))

    for i, img in enumerate(imgs):
        row_image.paste(img, (i * target_width_pixels, 0))

    row_images.append(row_image)

# Combine rows vertically
total_height = len(row_images) * target_height_pixels
final_width = 3 * target_width_pixels

final_image = Image.new('RGB', (final_width, total_height))

for i, row_img in enumerate(row_images):
    final_image.paste(row_img, (0, i * target_height_pixels))

# Save the final image
final_image.save("....", format='TIFF', dpi=(dpi, dpi))

# Optionally display
final_image.show()
```

```python
from PIL import Image

# Image paths
image_paths_list = [
    [
        "....",
        "....",
        "....",
        "...."
    ],
    [
        "....",
        "....",
        "....",
        "...."
    ]
]

# Allow large images
Image.MAX_IMAGE_PIXELS = None

# DPI and dimensions
dpi = 2400
original_width_inches = 2.5
base_width_pixels = int(original_width_inches * dpi * 0.8)  # Width: 80% of 2.5 inches at 2400 dpi
uniform_height_pixels = int(base_width_pixels * 0.75)       # Height: 75% of base width

# Vertical padding between rows
vertical_padding = 300

row_images = []
for image_paths in image_paths_list:
    imgs = []

    for path in image_paths:
        img = Image.open(path).convert("RGB")
        img = img.resize((base_width_pixels, uniform_height_pixels))
        imgs.append(img)

    # Create row image
    row_width = sum(img.width for img in imgs)
    row_height = uniform_height_pixels
    row_image = Image.new('RGB', (row_width, row_height), color=(255, 255, 255))

    # Paste images side by side
    x_offset = 0
    for img in imgs:
        row_image.paste(img, (x_offset, 0))
        x_offset += img.width

    row_images.append(row_image)

# Combine rows
final_width = max(row.width for row in row_images)
final_height = sum(row.height for row in row_images) + (len(row_images) - 1) * vertical_padding
final_image = Image.new('RGB', (final_width, final_height), color=(255, 255, 255))

y_offset = 0
for row_img in row_images:
    final_image.paste(row_img, (0, y_offset))
    y_offset += row_img.height + vertical_padding

# Save final image
final_image.save(
    "....",
    format='TIFF',
    dpi=(dpi, dpi)
)

# Display (optional)
final_image.show()
```
